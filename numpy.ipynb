```python
import numpy as np
```

1. arithmetic operators

```python
arr1=np.array([1,2,3])
arr2=np.array([3,2,4])
print(arr1+arr2)
print(arr1-arr2)
print(arr1*arr2)
print(arr1/arr2)
print(arr1**arr2)
print(arr1%arr2)
```

2. comparison operators


```python
arr1=np.array([1,2,3])
arr2=np.array([3,2,4])
print(arr1>arr2)
print(arr1==arr2)
print(arr1!=arr2)
```

3. ARRAY SHAPE OPERATIONS


```python
arr1=np.array([1,2,3])         #1dimentional array
arr2=np.array([[1,2,3],[5,4,6]])             #2dimentional array
arr3=np.array([[[1,2,3],[2,5,8],[7,8,9]]])   #3dimentional array
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)
```

5 dimentional array

```python
arr5=np.array([[[[[1,2,3],[2,3,4],[4,5,6],[2,4,8],[4,5,6]]]]])
print(arr5.shape)
```
