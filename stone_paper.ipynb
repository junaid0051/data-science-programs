```python
import random
print("welcome to stone paper")
print("Enter stone paper or scissor")
choices=["stone","paper","scissor",]
while True:
  user_choice=input("user_choice")
  computer_choice=random.choice(choices)
  print("computer choice:",computer_choice)



  if user_choice not in choices:
    print("Invalid user input!,enter valid input like stone paper scissor")
  continue

  if user_choice==computer_choice:
    print("draw")
  elif (
        (user_choice=="stone" and computer_choice=="scissor") or
        (user_choice=="paper" and computer_choice=="stone") or
        (user_choice=="scissor" and computer_choice=="paper")
       ):
        print("you win!")
  else:
     print("computer wins!")

  play_again=input("Do you want to continue the game :yes/no")
  if play_again!="yes":
    print("thankyou for playing ")
    break


```
