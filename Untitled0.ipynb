```python
print("guess my number")
print("the number between 1 to 100")
secret_number=26
for i in range(0,8):
    user_input=int(input("enter the guess number:"))
    if secret_number==user_input:
       print("congratulation, you are enter correct number")
       print(f"congratulation you have guessed the number {secret_number} in {i} attempts")
       break
    elif user_input>100:
       print("invalid number")
    elif user_input<1:
       print("invalid number")
    elif secret_number<user_input:
       print("this number is greater than secret number")
    elif secret_number>user_input:
       print("this number is less than secret number")
    else:

       print("sorry, you are enter a wrong number")
```
