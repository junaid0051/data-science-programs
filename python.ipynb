```python


# HERE ME CREATE PARENT CLASS


class animal:
    def eat(self):
        print("this animal eat food")

#AFTER THE CHILD CLASS


class dog(animal):
    def bark(self):
        print("dog barks")


#AFTER THE PARENT , CHILD CLASS OBJECT CREATION


d=dog()
d.eat()       #INHERITED FROM ANIMAL
d.bark()        #DOGS OWN METHOD
```

```python
class user:
   def login(self):
    print("user are loging")


class student(user):
   def switch_video(self):
    print("student watched the video")


class teacher(user):
   def upload_video(self):
    print("teacher upload a video")



s=student()
s.login()
s.switch_video()
t=teacher()
t.login()
t.upload_video()
```

```python
class father:
  def skill1(self):
    #method of father class
    print("fathers skill:gardening")



class mother:
  def skill2(self):
    #method of mother class
    print("mothers skill:cooking")



class child(father,mother):
  #child inherits from both father and mother
  pass


#creating object of child class
c=child()
c.skill1()
c.skill2()
```

```python
#grandparent class
class grandparent:
   def display(self):
     print("grandparent class")


#parent class inherits from grandparent
class parent(grandparent):
  def show(self):
     print("parent class")


#child class inherits from parent
class child(parent):
  def see(self):
     print("child class")



#create object of child
obj = child()
obj.display()  #method from grandparents
obj.show()     #method from parent
obj.see()      #method from child
```

```python
#base class
class animal:
  def eat(self):
    print("animal eats")


#derived class (single inhertance)
class mammal(animal):
  def walk(self):
     print("mammal walks")



#another base class
class pet:
  def play(self):
     print("pet plays")


#derived class (multiple inhertance)
class dog(mammal,pet):
  def bark(self):
    print("dog barks")



#creating object
d=dog()
#accessing method
d.eat()     #from animal
d.walk()    #from mammal
```
